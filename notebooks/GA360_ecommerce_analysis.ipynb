## 一、项目背景

互联网电商经营中，业务团队通常需要同时回答三个问题：**流量从哪里来、用户在哪里流失、哪些用户更有价值**。本项目围绕这三类问题，选取 Google Analytics 360 Sample（Google Merchandise Store 的真实匿名化电商数据）作为分析对象，从用户进入网站到完成购买的全链路行为入手，构建一套可复现、统计口径一致的经营分析框架。

数据覆盖一整年（2016-08-01 至 2017-08-01），包含 **903,653 个 Session** 与 **596,585 条关键电商行为事件**，足以支撑渠道、漏斗、用户生命周期三个层面的经营诊断。

需要说明的是，本报告采用 **观察性分析** 视角：目标是识别现象、量化环节、并提出可验证的运营假设，而非证明某项策略与结果之间的因果关系。

## 执行摘要

本报告基于一整年的 GA360 电商数据，对渠道、漏斗与用户价值进行诊断，得到以下核心结论。

### 1. 经营基线

观察期内共有 **714,167 名 UV**、**903,653 个 Session**，其中购买 Session 为 **11,552** 个，Session 购买率（Session 层面）为 **1.28%**，总收入约为 **$1,540,071.24**。

### 2. 渠道差异

- **Referral（引荐）**：购买率约 **5.08%**，收入约 **$651,429.91**，是观察期内效率与收入贡献都较为突出的渠道。
- **Social（社交）**：购买率仅约 **0.05%**，跳出率约 **65.24%**，需要重点检查流量质量与落地页匹配度。
- **Organic Search（自然搜索）**：贡献最大的流量规模（311,607 UV），但购买率处于中低水平，属于典型的“大流量、低效率”渠道。

### 3. 新老用户差异

- 回访用户 Session 购买率 **3.53%**，明显高于新用户的 **0.64%**。
- 回访用户贡献约 **$1,157,745.28** 收入，说明复访与留存用户的商业效率远高于一次性拉新。

### 4. 漏斗流失

严格顺序漏斗显示，最大流失发生在 **浏览 → 加购**（流失率 **64.53%**），其次是 **加购 → 结账**（流失率 **61.20%**）。因此，第一优先级应放在商品详情页的信息表达、加购引导，以及购物车到结账的流程优化，而不是只关注最终支付环节。

### 5. 用户价值分层

RFM 分层共覆盖 **10,022 名购买用户**，累计订单收入约 **$1,627,429.54**。其中识别出 **385 名高价值用户**，约 **3.84%** 的人数贡献了 **25.24%** 的收入，人均消费最高，应优先维护、避免过度依赖优惠；**风险用户**（2,281 人，收入贡献 12.40%）则是召回策略的重点对象。

### 6. 复合洞察

把渠道、用户类型、设备与用户价值叠加，得到最具操作性的结论：

- **高价值用户高度集中于特定组合**：356 名高价值用户中，**约 55% 来自 Referral、97% 使用桌面端、90% 是首次进入的新用户**——高价值不是老客买得多，而是对的人在优质渠道一次性买得贵。
- **渠道决定买不买、新老决定买多少、设备是放大器**：Referral 新用户购买率 6.91% 甚至高于 Direct 回访用户 4.61%；而桌面端回访购买率是移动端的约 3.3 倍。**价值 ≈ 渠道（意愿）× 新老（强度）× 设备（放大）**。

### 分析边界

以上结论均来自观察性数据。渠道差异与漏斗流失只能说明现象，不能直接证明因果关系；页面、渠道与运营策略的效果，需要通过独立 A/B Test 或准实验进一步验证。

### 1.2.1 构造候选 Session 标识

**分析目标**

使用用户标识和访问标识构造候选 `session_id`，并评估它是否能够唯一识别一次用户访问。

**为什么需要这一步**

`fullVisitorId` 只能识别用户，不能区分同一用户的多次访问；`visitId` 也不能脱离用户标识单独使用。因此先构造：

```text
session_id = fullVisitorId + "_" + visitId
```

**对后续分析的作用**

候选 `session_id` 将用于连接 `Session` 信息与漏斗行为。只有先检查它的唯一性，才能判断是否需要加入日期等额外字段，避免跨日期错误合并访问。

````python
sessions["session_id"] = (
    sessions["fullVisitorId"].astype("string")
    + "_"
    + sessions["visitId"].astype("string")
)

funnel_hits["session_id"] = (
    funnel_hits["fullVisitorId"].astype("string")
    + "_"
    + funnel_hits["visitId"].astype("string")
)

print("Session表唯一 session_id 数:",
      sessions["session_id"].nunique())

print("Session表记录数:",
      len(sessions))

print("Session表重复 session_id 数:",
      sessions["session_id"].duplicated().sum())

print("漏斗表唯一 session_id 数:",
      funnel_hits["session_id"].nunique())

print("漏斗表记录数:",
      len(funnel_hits))

print("漏斗表重复 session_id 数:",
      funnel_hits["session_id"].duplicated().sum())
````

### 1.2.2 候选 Session 标识的重复审计

**分析目标**

检查重复 `session_id` 是否只是完全重复记录，还是同一候选键对应了不同日期、渠道或交易信息。

**为什么需要这一步**

`Session` 表理论上一行代表一次访问，因此重复候选键需要解释。完全重复可以安全去重；字段存在冲突时，直接删除可能丢失真实访问，直接保留又会重复计算经营指标。

**本节判断标准**

- `record_count`：同一候选 Session 出现的记录数。
- `different_dates`：是否跨多个日期。
- `different_channels`：渠道字段是否冲突。
- `different_transactions`：交易次数是否冲突。
- `different_revenue`：交易收入是否冲突。

**对后续分析的作用**

审计结果将决定最终 Session Key 的组成方式和 `Session` 表是否需要去重，从而保证渠道、交易和收入指标的准确性。

````python
import duckdb

duplicate_summary_sql = """
WITH duplicate_sessions AS (
    SELECT
        session_id,
        COUNT(*) AS record_count,
        COUNT(DISTINCT fullVisitorId) AS different_users,
        COUNT(DISTINCT date) AS different_dates,
        COUNT(DISTINCT channelGrouping) AS different_channels,
        COUNT(DISTINCT transactions) AS different_transactions,
        COUNT(DISTINCT transactionRevenue) AS different_revenue
    FROM sessions
    GROUP BY session_id
    HAVING COUNT(*) > 1
)

SELECT
    session_id,
    record_count,
    different_users,
    different_dates,
    different_channels,
    different_transactions,
    different_revenue
FROM duplicate_sessions
ORDER BY session_id
LIMIT 10
"""

duplicate_summary_sql_result = duckdb.sql(
    duplicate_summary_sql
).df()

display(duplicate_summary_sql_result)
````

### 1.2.3 冲突样本核查

**分析目标**

抽取一个存在日期冲突的候选 `session_id`，查看其原始记录，验证重复键到底代表重复数据还是不同日期的访问。

**为什么需要这一步**

汇总统计只能告诉我们“存在冲突”，不能说明冲突的业务含义。查看具体样本可以确认 `date`、`visitStartTime`、渠道和交易字段是否共同变化。

**对后续分析的作用**

该案例将帮助我们选择最终 Session Key。如果同一用户和 `visitId` 跨日期出现，就必须把日期纳入键，避免把不同日期的访问错误合并。

````python
conflict_session_id = duplicate_summary_sql_result.loc[
    duplicate_summary_sql_result["different_dates"] > 1,
    "session_id",
].iloc[0]

conflict_records = (
    sessions.loc[
        sessions["session_id"] == conflict_session_id,
        [
            "fullVisitorId",
            "visitId",
            "date",
            "visitStartTime",
            "channelGrouping",
            "source",
            "medium",
            "transactions",
            "transactionRevenue",
        ],
    ]
    .sort_values(["date", "visitStartTime"])
)

display(conflict_records)
````

#### 1.2.4 冲突样本结论与 Session Key 修正

案例显示，同一个 `fullVisitorId + visitId` 组合可能对应不同的 `date` 和 `visitStartTime`。因此，候选 `session_id` 在本次导出数据中不能完全保证唯一。

为避免跨日期错误合并，最终采用：

```text
session_key = fullVisitorId + "_" + date + "_" + visitId
```

**对后续分析的作用**

最终 `session_key` 同时用于 `Session` 表和漏斗事件表：

- `Session` 表中应一行对应一个唯一 `session_key`。
- 漏斗事件表中允许同一个 `session_key` 出现多行，因为一个 Session 可以包含多个行为事件。
- 后续漏斗分析将以 `session_key` 为聚合单位，而不是以事件行数作为人数。

````python
sessions["session_key"] = (
    sessions["fullVisitorId"].astype("string")
    + "_"
    + sessions["date"].astype("string")
    + "_"
    + sessions["visitId"].astype("string")
)

funnel_hits["session_key"] = (
    funnel_hits["fullVisitorId"].astype("string")
    + "_"
    + funnel_hits["date"].astype("string")
    + "_"
    + funnel_hits["visitId"].astype("string")
)

print("Session表 session_key 唯一数:",
      sessions["session_key"].nunique())

print("Session表记录数:",
      len(sessions))

print("Session表 session_key 重复记录数:",
      sessions["session_key"].duplicated().sum())

print("漏斗表 session_key 唯一数:",
      funnel_hits["session_key"].nunique())

print("漏斗表 session_key 重复记录数:",
      funnel_hits["session_key"].duplicated().sum())
````

### 1.3 审计漏斗行为编码

**本节目的：** 确认漏斗事件表中的 `actionType` 是否包含项目关注的四类电商行为，并统计每类行为的事件数量。

**当前项目的漏斗口径：**

```text
商品详情浏览 → 加入购物车 → 进入结账 → 完成购买
```

对应编码为：

| `actionType` | 行为 |
|---:|---|
| `2` | 商品详情浏览 |
| `3` | 加入购物车 |
| `5` | 进入结账 |
| `6` | 完成购买 |

**为什么做：** 代码中的数字编码只是原始数据字段，不能未经检查就直接当作业务行为使用；同时，事件行数也不能直接当作用户数或 Session 数。

**对下一步的作用：** 确认编码后，下一节才能按 `session_key` 聚合事件，构建严格的行为顺序漏斗，并计算各阶段的 Session 数、转化率和流失率。

````python
action_type_summary = (
    funnel_hits["actionType"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("actionType")
    .reset_index(name="event_count")
)

display(action_type_summary)

expected_action_types = {2, 3, 5, 6}
observed_action_types = set(
    funnel_hits["actionType"]
    .dropna()
    .astype(int)
    .unique()
)

print("实际 actionType:", sorted(observed_action_types))
print("项目关注的 actionType:", sorted(expected_action_types))
print("缺少的关键行为:", sorted(expected_action_types - observed_action_types))
print("额外行为编码:", sorted(observed_action_types - expected_action_types))
````

#### 1.3.1 审计结果解释

实际观察到的关键行为编码为 `[2, 3, 5, 6]`，与项目预设口径完全一致：

- 商品详情浏览事件：362,607 条
- 加入购物车事件：121,427 条
- 进入结账事件：88,019 条
- 完成购买事件：24,532 条
- 缺少的关键行为：无
- 额外行为编码：无

这里的 `event_count` 是行为事件行数，不是用户数或 Session 数。同一 Session 可能产生多条相同或不同的行为事件，因此下一步必须按 `session_key` 去重聚合后再计算漏斗阶段规模。

### 2.1 标准化 Session 表字段

**分析目标**

将原始 Session 表转换为可直接用于经营分析的指标底表，统一日期、时间、收入、交易、新老用户和跳出字段的口径。

**为什么需要这一步**

原始字段中，日期是 `YYYYMMDD` 数字，访问时间是 Unix 时间戳，收入以微单位保存；交易、收入、跳出和新用户字段还存在缺失值。若不先标准化，后续渠道比较和趋势分析容易出现口径不一致。

**本节处理规则**

- `date` 转换为分析日期 `analysis_date`。
- `visitStartTime` 转换为 Session 开始时间 `session_start`。
- `transactionRevenue` 除以 1,000,000，转换为美元收入 `revenue_usd`。
- 缺失交易次数和收入按无交易处理。
- `newVisits = 1` 标记为新用户，其余记录归为回访用户。
- `bounces = 1` 标记为跳出 Session。

**对下一步的作用**

本节完成后，才能按日期、渠道、设备和用户类型计算 UV、Session 数、购买率、收入和跳出率。

````python
import numpy as np

sessions_analysis = sessions.copy()

sessions_analysis["analysis_date"] = pd.to_datetime(
    sessions_analysis["date"].astype("string"),
    format="%Y%m%d",
)

sessions_analysis["session_start"] = pd.to_datetime(
    sessions_analysis["visitStartTime"],
    unit="s",
)

sessions_analysis["revenue_usd"] = (
    sessions_analysis["transactionRevenue"]
    .fillna(0)
    / 1_000_000
)

sessions_analysis["transaction_count"] = (
    sessions_analysis["transactions"]
    .fillna(0)
)

sessions_analysis["is_new_user"] = (
    sessions_analysis["newVisits"] == 1
)

sessions_analysis["user_type"] = np.where(
    sessions_analysis["is_new_user"],
    "新用户",
    "回访用户",
)

sessions_analysis["has_transaction"] = (
    sessions_analysis["transaction_count"] > 0
)

sessions_analysis["is_bounce"] = (
    sessions_analysis["bounces"]
    .fillna(0)
    .eq(1)
)

display(
    sessions_analysis[
        [
            "fullVisitorId",
            "session_key",
            "analysis_date",
            "session_start",
            "user_type",
            "transaction_count",
            "revenue_usd",
            "has_transaction",
            "is_bounce",
        ]
    ].head()
)
````

### 2.2 总体经营指标基线

**分析目标**

从标准化 Session 表计算全站总体的用户数、Session 数、购买 Session 数、购买率、交易数、收入和跳出率，建立后续分组比较的参照水平。

**为什么需要全站基线？**

分组指标必须放在全站总体水平下解释。全站基线提供统一参照，使我们能够判断某个渠道、用户类型或设备的表现是高于、接近还是低于整体水平。

同时，业务判断不能只看购买率，还要结合 UV、Session 数、购买 Session 数、交易数、收入和跳出率。高购买率但规模很小的分组，和购买率一般但贡献大量收入的分组，运营价值并不相同。

因此，本项目对分组表现采用“规模 + 效率 + 商业产出”的综合判断，而不是只根据单一指标下结论。

**本节指标口径**

- UV(Unique Visitor)：去重后的 `fullVisitorId` 数量。
- Session 数：去重后的 `session_key` 数量。
- 购买 Session 数：`has_transaction = True` 的 Session 数量。
- 购买率：购买 Session 数除以 Session 总数。
- 交易数：`transaction_count` 的总和。
- 总收入：`revenue_usd` 的总和，单位为美元。
- 跳出率：跳出 Session 数除以 Session 总数。

**对下一步的作用**

本节生成全站经营基线。下一步将使用相同的分子、分母和字段口径，按渠道、新老用户和设备拆分指标，避免分组之间出现不可比的统计口径。

````python
overall_metrics = pd.DataFrame(
    {
        "metric": [
            "UV(Unique Visitor)",
            "Session 数",
            "购买 Session 数",
            "购买率",
            "交易数",
            "总收入（美元）",
            "跳出率",
        ],
        "value": [
            sessions_analysis["fullVisitorId"].nunique(),
            sessions_analysis["session_key"].nunique(),
            int(sessions_analysis["has_transaction"].sum()),
            sessions_analysis["has_transaction"].mean(),
            sessions_analysis["transaction_count"].sum(),
            sessions_analysis["revenue_usd"].sum(),
            sessions_analysis["is_bounce"].mean(),
        ],
    }
)

display(overall_metrics)
````

### 2.3 渠道经营表现分析

**分析目标**

按 `channelGrouping` 拆分 Session 表，比较不同渠道的用户规模、Session 规模、购买效率、交易贡献、收入贡献和跳出表现。

**为什么需要这一步**

总体指标只能说明全站发生了什么，不能回答流量从哪里来、哪些渠道带来的用户更有购买效率。渠道拆分可以帮助业务识别“高流量渠道”和“高效率渠道”之间是否存在差异。

**本节指标口径**

- UV：渠道内去重 `fullVisitorId` 数量。
- Session 数：渠道内去重 `session_key` 数量。

#### 2.4.1 结果解读与综合洞察

本节基于 `source_medium_user_type_daily` 的交叉数据，从渠道、用户类型、时间三个维度做推断性分析。以下结论来自观察性数据，用于提出运营假设，不直接证明因果关系。

1. **新老用户结构：留存用户的商业价值远高于拉新**

- 回访用户以 **22.2%** 的流量贡献了 **75.2%** 的收入（$\$1{,}157{,}745.28$），购买率 **3.53%**；<br>新用户占 **77.8%** 流量却只贡献 **24.8%** 收入（$\$382{,}325.96$），购买率 **0.64%**。
- **推断**：回访用户购买率是新用户的 5.5 倍，叠加更高客单价与复购频次，使其收入贡献被放大到四分之三。新用户占了近八成流量却只换回四分之一收入——**增长的重心应从"拉新"转向"让新用户变成回访用户"**。把一名新用户推向复购，比从零拉一个新用户更划算。

2. **渠道效率与规模是两套逻辑，且差距悬殊**

| 渠道 | 流量占比 | 收入占比 | 效率倍数 |
|---|---:|---:|---:|
| Referral | 11.6% | **42.3%** | **3.6×** |
| Direct | 15.8% | 28.2% | 1.8× |
| Organic Search | 42.2% | 21.2% | 0.5× |
| Social | 25.0% | **0.3%** | 0.01× |

- **推断**：同样一个 Session，Referral 带来的收入约为 Social 的 **130~300 倍**。这背后是用户意图的本质差异——Referral 用户带着明确购买意向，Social 用户是刷内容时被吸引、并未打算购买。
- **洞察**：Social 应作为"规模 / 曝光"渠道而非"购买"渠道考核；**Referral 才是利润核心**——仅 11.6% 流量却贡献 42.3% 收入，值得加码。

3. **渠道 × 用户类型交叉：识别哪些渠道拉来的是真买家**

| 渠道 | 新用户购买率 | 回访购买率 | 提升倍数 |
|---|---:|---:|---:|
| Referral | 2.87% | **7.36%** | 2.6× |
| Direct | 0.91% | 3.00% | 3.3× |
| Social | 0.02% | 0.35% | 17× |

- **推断**：Referral 拉来的新用户购买意愿本已较强（2.87%），复购后更升至 7.36%——**Referral 带来的是高质量种子用户**，值得做"拉新→复购"激活；Social 新用户几乎不买（0.02%），即使成为回访也仅 0.35%——**Social 拉来的是低意图泛流量**，做复购激活是浪费预算。

4. **风险：收入高度依赖单一来源（归因可靠性存疑）**

- Referral 总收入 $\$651{,}429.91$ 中，**`(direct)/(none)` 单一组合占约 94.5%**（$\$615{,}391.69$），其余所有来源合计仅约 $\$36{,}000$。
- **推断**：所谓"推荐渠道健康"，实为**对单一来源的过度依赖**；且该来源标签为 `source=(direct)`、`medium=(none)`，实为无法归因的"自引 / 直接"流量（可能混入品牌词搜索、书签、直链）。因此"Referral 高效"这一结论的**归因可靠性需要谨慎对待**，它更像"高意图的直接/品牌流量"，而非纯推荐流量。

5. **时间趋势：峰值是促销脉冲，而非流量驱动**

- 收入最高的几天为 **2017-04-05（$\$27{,}151$）**、**2017-02-14（情人节，$\$23{,}028$）**、2016-09-16、2016-08-25 等；但 session 与收入相关系数仅 **0.349**（弱相关）。
- **推断**：这些收入尖峰并非靠流量拉起（session 未同步暴涨），而是**单日客单价被活动/促销拉高**的典型特征——例如 2017-02-14 仅 2,379 个 session 就产生 $\$23{,}028$。说明**靠自然流量增长已边际递减**，应把杠杆转向**活动策划与高价值用户定向**。

**综合结论（一句话）**

> 本业务的真正金矿是 **"Referral 里的留存用户"**，而非最大的流量。七成多流量（Social + 新用户）换不回对应收入；而 22% 流量的回访用户 + 11.6% 流量的 Referral 贡献了 75% + 42% 的收入。**核心策略**：加码 Referral、激活复购、用活动而非流量驱动收入，同时警惕 Referral 对单一来源的过度依赖。

（注：本节结论基于观察性交叉数据，用于提出可验证的运营假设；页面、渠道与运营策略的效果需通过独立 A/B Test 或准实验确认。）


## 3. 电商行为路径与转化漏斗

本章将漏斗事件从事件行转换为 Session 层面的行为路径，计算商品浏览、加入购物车、进入结账和完成购买之间的阶段规模、转化率与流失率。

**本章分析顺序：**

1. 先按 `session_key` 汇总每个行为是否出现，建立宽口径行为矩阵。
2. 再按 `hitNumber` 检查行为发生顺序，构建严格顺序漏斗。
3. 最后按渠道、新老用户和设备拆分漏斗，定位不同人群的流失节点。

**核心原则：** 漏斗阶段的分母和人数使用去重后的 `session_key`，不能直接使用事件行数。

### 3.1 Session 行为标记矩阵

**分析目标**

将漏斗事件表转换为 Session 层面的宽表：每个 `session_key` 一行，并标记该 Session 是否出现浏览、加购、结账和购买行为。

**为什么需要这一步**

同一 Session 可能包含多条事件记录。直接统计事件行数会重复计算同一个 Session；先按 Session 聚合，才能得到正确的阶段规模。

**对下一步的作用**

宽口径矩阵可以先回答“有多少 Session 曾经到达过每个阶段”。下一步再使用 `hitNumber` 检查顺序，识别真正完成连续行为路径的 Session。

````python
stage_map = {
    2: "浏览",
    3: "加购",
    5: "结账",
    6: "购买",
}

funnel_stage_matrix = (
    funnel_hits.loc[
        funnel_hits["actionType"].isin(stage_map),
        ["session_key", "actionType"],
    ]
    .drop_duplicates()
    .assign(stage_reached=True)
    .pivot_table(
        index="session_key",
        columns="actionType",
        values="stage_reached",
        aggfunc="max",
        fill_value=False,
    )
    .rename(columns=stage_map)
    .reset_index()
)

for stage_name in stage_map.values():
    if stage_name not in funnel_stage_matrix.columns:
        funnel_stage_matrix[stage_name] = False

funnel_stage_matrix = funnel_stage_matrix[
    ["session_key", "浏览", "加购", "结账", "购买"]
]

funnel_stage_matrix.to_csv(
    output_table_dir / "funnel_stage_matrix.csv",
    index=False,
    encoding="utf-8-sig",
)

display(funnel_stage_matrix.head())

print("行为矩阵 Session 数:", len(funnel_stage_matrix))
````

### 3.2 构建严格顺序漏斗

**分析目标**

按每个 Session 内的 `hitNumber` 对行为排序，判断用户是否依次完成：

```text
商品浏览 → 加入购物车 → 进入结账 → 完成购买
```

**为什么需要这一步**

3.1 的宽口径矩阵只判断某个行为是否出现过，无法确认行为顺序。例如同一个 Session 可能同时出现购买和浏览，但这不代表用户按照正常路径完成了漏斗。

**严格口径**

- 进入加购阶段：该 Session 先出现浏览，再出现加购。
- 进入结账阶段：该 Session 先出现浏览、加购，再出现结账。
- 完成购买阶段：该 Session 先出现浏览、加购、结账，再出现购买。

**对下一步的作用**

严格漏斗将提供可解释的阶段人数、相邻转化率和流失率，并可进一步按渠道、新老用户和设备拆分。

````python
ordered_actions = (
    funnel_hits
    .sort_values(["session_key", "hitNumber"])
    .groupby("session_key", sort=False)["actionType"]
    .agg(list)
)


def get_strict_funnel_progress(action_types):
    expected_sequence = [2, 3, 5, 6]
    progress = 0
    flags = [False, False, False, False]

    for action_type in action_types:
        if progress < len(expected_sequence):
            if action_type == expected_sequence[progress]:
                flags[progress] = True
                progress += 1

    return pd.Series(
        {
            "浏览": flags[0],
            "加购": flags[1],
            "结账": flags[2],
            "购买": flags[3],
        }
    )


strict_funnel = ordered_actions.apply(
    get_strict_funnel_progress
).reset_index()

strict_funnel.to_csv(
    output_table_dir / "strict_funnel_session_flags.csv",
    index=False,
    encoding="utf-8-sig",
)

display(strict_funnel.head())
print("严格漏斗 Session 数:", len(strict_funnel))
````

### 3.3 计算严格漏斗阶段规模与流失率

**分析目标**

将严格顺序标记汇总为漏斗阶段指标，计算各阶段的 Session 数、相邻阶段转化率和流失率。

**漏斗起点口径**

当前漏斗事件表只包含关键电商行为，因此本节以发生商品详情浏览的 Session 作为漏斗起点，不把所有网站 Session 直接当作商品漏斗的浏览分母。

**指标定义**

- 阶段 Session 数：满足该阶段严格顺序条件的去重 `session_key` 数量。
- 阶段转化率：下一阶段 Session 数除以上一阶段 Session 数。
- 阶段流失率：$1 -$ 阶段转化率。

**对下一步的作用**

本节将定位整体漏斗中流失最明显的阶段。后续可以按渠道、新老用户和设备拆分同一套漏斗指标，判断不同人群的流失位置是否不同。

````python
funnel_stage_order = ["浏览", "加购", "结账", "购买"]

funnel_stage_counts = strict_funnel[funnel_stage_order].sum().astype(int)

funnel_summary = pd.DataFrame(
    {
        "stage": funnel_stage_order,
        "session_count": [
            funnel_stage_counts[stage]
            for stage in funnel_stage_order
        ],
    }
)

funnel_summary["conversion_rate"] = np.nan
funnel_summary["dropoff_rate"] = np.nan

for index in range(1, len(funnel_summary)):
    previous_count = funnel_summary.loc[index - 1, "session_count"]
    current_count = funnel_summary.loc[index, "session_count"]
    conversion_rate = current_count / previous_count

    funnel_summary.loc[index, "conversion_rate"] = conversion_rate
    funnel_summary.loc[index - 1, "dropoff_rate"] = 1 - conversion_rate

funnel_summary.loc[0, "conversion_rate"] = 1.0
funnel_summary.loc[len(funnel_summary) - 1, "dropoff_rate"] = np.nan

funnel_summary.to_csv(
    output_table_dir / "overall_funnel.csv",
    index=False,
    encoding="utf-8-sig",
)

display(funnel_summary)
````

#### 3.3.1 漏斗指标结果解读与分析

**字段口径**

`funnel_summary` 中每一行代表一个漏斗阶段：

- `session_count`：严格到达该阶段的 Session 数。
- `conversion_rate`：相邻阶段转化率（从上一阶段进入当前阶段的比例）。
- `dropoff_rate`：相邻阶段流失率（从上一阶段到当前阶段流失的比例）。

第一阶段“浏览”为漏斗起点，转化率固定为 **100%**；最后阶段“购买”无下一阶段，流失率为空。

**整体漏斗结果**

| 阶段 | Session 数 | 占起点比例 | 相邻转化率 | 流失率 |
|---|---:|---:|---:|---:|
| 浏览 | 124,188 | 100.0% | — | — |
| 加购 | 44,044 | 35.5% | 35.47% | **64.53%** |
| 结账 | 17,088 | 13.8% | 38.80% | **61.20%** |
| 购买 | 9,072 | 7.3% | **53.09%** | 46.91% |

**推断与洞察**

- **最大流失发生在「浏览 → 加购」（流失 64.53%）**：超过 **6 成** 浏览过商品详情的用户没有加入购物车，这是最大的瓶颈。相比之下，「结账 → 购买」仍有 **53%** 通过率。**这推翻了"用户卡在支付环节"的直觉——真正的问题在于商品页无法把"看一眼"变成"想买"**。说明商品详情页在价格/促销表达、信任信息、加购按钮可见性、或商品吸引力上存在短板。
- **「浏览 → 结账」仅 13.8%**：从起点到结账前被筛掉 **86%**，说明大部分用户在商品浏览与购物车阶段就已流失，少有人走到结账。
- **每加购 Session 的潜在价值**：全站收入 $\$1{,}540{,}071$ 对应 44,044 个加购 Session，约 **$\$35$/加购**。这意味着：**若能把「浏览→加购」的 64.53% 流失哪怕压缩一部分，每一分转化都直接对应约 $35 的收入增量**，ROI 极高。
- **优化优先级（基于流失率排序）**：
  1. **浏览 → 加购**（流失 64.53%）：商品详情页信息完整性、价格与促销表达、用户信任元素、加购按钮可见性。
  2. **加购 → 结账**（流失 61.20%）：购物车费用展示、配送说明、登录要求、结算步骤阻力。
  3. **结账 → 购买**（流失 46.91%）：支付流程、异常报错、移动端体验。

**分析边界**

以上为观察性漏斗结果，能定位流失环节并提出优化假设，但不能证明某项页面改动必然导致转化变化——后续需通过独立 A/B Test 或准实验验证。

**提示**：解释漏斗时应优先关注**流失率最高的相邻阶段**（这里是浏览→加购），而非只看最终购买率（7.3%）。


### 3.4 渠道与用户类型漏斗拆分

**分析目标**

将严格漏斗阶段标记与 Session 表中的 `channelGrouping`、`user_type` 和 `deviceCategory` 关联，比较不同人群在浏览、加购、结账和购买阶段的流失差异。

**为什么需要这一步**

整体漏斗只能告诉我们全站最严重的流失环节，不能说明哪些渠道或用户类型的问题更突出。分组漏斗可以帮助业务区分：

- 是某个渠道带来的流量质量较低；
- 还是新用户在早期阶段流失更多；
- 还是某类设备在结账或购买阶段表现较弱。

**关键口径**

Session 属性表先按 `session_key` 保留一行，再与严格漏斗标记合并；漏斗阶段人数继续使用去重后的 Session 数，不使用事件行数。

**对下一步的作用**

本节将生成渠道、用户类型和设备维度的漏斗结果，为 Tableau 联动筛选和具体运营策略提供依据。

````python
session_attributes = (
    sessions_analysis[
        [
            "session_key",
            "channelGrouping",
            "user_type",
            "deviceCategory",
        ]
    ]
    .drop_duplicates("session_key")
)

funnel_with_attributes = strict_funnel.merge(
    session_attributes,
    on="session_key",
    how="left",
    validate="one_to_one",
)

def summarize_funnel_by(group_column):
    grouped = (
        funnel_with_attributes
        .groupby(group_column)[funnel_stage_order]
        .sum()
        .reset_index()
    )

    rows = []

    for _, group_row in grouped.iterrows():
        group_value = group_row[group_column]
        counts = [int(group_row[stage]) for stage in funnel_stage_order]

        for stage_index, stage in enumerate(funnel_stage_order):
            if stage_index == 0:
                conversion_rate = 1.0
                dropoff_rate = 0.0
            else:
                conversion_rate = counts[stage_index] / counts[stage_index - 1]
                dropoff_rate = 1 - conversion_rate

            rows.append(
                {
                    group_column: group_value,
                    "stage": stage,
                    "session_count": counts[stage_index],
                    "conversion_rate": conversion_rate,
                    "dropoff_rate": dropoff_rate,
                }
            )

    return pd.DataFrame(rows)

channel_funnel = summarize_funnel_by("channelGrouping")
user_type_funnel = summarize_funnel_by("user_type")
device_funnel = summarize_funnel_by("deviceCategory")

channel_funnel.to_csv(
    output_table_dir / "channel_funnel.csv",
    index=False,
    encoding="utf-8-sig",
)

user_type_funnel.to_csv(
    output_table_dir / "user_type_funnel.csv",
    index=False,
    encoding="utf-8-sig",
)

device_funnel.to_csv(
    output_table_dir / "device_funnel.csv",
    index=False,
    encoding="utf-8-sig",
)

display(channel_funnel.head(12))
display(user_type_funnel)
display(device_funnel)
````

#### 3.4.1 分组漏斗结果解释

- `channel_funnel` 用于比较不同渠道在哪个阶段流失。
- `user_type_funnel` 用于比较新用户和回访用户的行为路径差异。
- `device_funnel` 用于识别不同设备上的漏斗阶段差异。
- 分组结果仍应同时参考 `session_count` 和 `dropoff_rate`，避免小样本分组的极端比例被误判为稳定规律。

这些结果可用于提出运营假设，例如优化高流量渠道的加购引导，或针对新用户强化商品介绍和首次购买引导；它们不能单独证明某项策略的因果效果。

### 组合层面：三根杠杆如何叠加（核心）

**单一维度看（渠道/用户/设备各自看）会掩盖真相，真正的洞察来自它们的交叉组合。** 下表列出经过样本量验证的显著组合：

| 组合 | 浏览 | 最终转化 | 解读 |
|---|---:|---:|---|
| **Referral×回访×desktop** | 14,791 | **20.42%** | **全矩阵最强组合**：加购率 51.68%（最高）、结账→购买 67.97%。三轴同命中（信任渠道×复购关系×桌面深思决策） |
| Organic Search×新用户×desktop | 31,354 | 3.54% | **全站最大单一流量池（占浏览 25%）**，却只产出 12% 购买，最大"未激活蓝海" |
| Social×新用户×mobile | 1,227 | 0.16% | 三重负向叠加（低意图渠道×无信任新客×最差分机），几乎不成交 |
| Affiliates×新用户×desktop | 1,212 | **0.17%** | 加购率 37.6% 看似正常，但结账→购买仅 3.57%——**"伪装优质"的弃单渠道** |

### 结构规律（跨组合稳定成立）

1. **用户类型是最普遍、最稳定的杠杆**：同一渠道同设备下，回访用户转化普遍为新用户的 **1.8~2.7 倍**（Organic 2.72×、Direct 2.68×；Paid 1.78×、Referral 1.87×），几乎在成规模的组合中无例外。它不是幅度最大的杠杆，但**在几乎每个组合都成立**。
2. **设备是稳定分层因子**：desktop 显著优于 mobile（同渠道同用户下约 1.7~2×以上，主流渠道数千至数万浏览、方向一致），移动端呈"**大流量、弱转化**"（如 Organic×新用户×mobile 浏览 1.3 万、转化 1.68%），是典型的**浏览/入口设备而非成交场景**。
3. **渠道优势是条件性的（有边界）**：Referral 在 **desktop** 称王（回访 20.4%），但在 mobile 无规模（仅 97/476 浏览），移动端几乎所有渠道购买量都极小（个位数），**"Referral 全面最好"只在桌面端成立**，不能外推到移动端。
4. **关键修正（重要）**：绝对的放大倍率在不同组合有强异质性，不能一律说成"2~4 倍"。Social 的新→老倍数高达 5.67×、但基数值极低；Display 是新老转化**几乎相等**的少数反例（0.080 vs 0.082）——Display 拉来的新客本身就有价值，缺的是量不是转化。

### 最有价值的三个组合方向（含绝对增量）

**基于"若把某组合提升到同渠道回访基准"估算的绝对可撬动订单（浏览×差距）：**

| 组合 | 现转化 | 同渠道回访基准 | 若达基准约可多卖 |
|---|---:|---:|---:|
| Organic Search×新用户×desktop | 3.54% | 9.61% | **≈ +1,905 单** |
| Referral×新用户×desktop | 10.93% | 20.42% | ≈ +1,115 单 |
| Direct×新用户×desktop | 5.11% | 13.76% | ≈ +845 单 |
| Paid Search×新用户×desktop | 7.21% | 12.82% | ≈ +105 单 |
| Social×新用户×desktop | 0.70% | 3.44% | ≈ +70 单 |

> **结论**：绝对增量最大的是 **Organic Search×新用户×desktop（约 1,905 单）**——它基数最大、落差最大，是最值得投入复购激活/再营销的池子。Referral、Direct 次之。而 **Social（0.70%）即便提升 5 倍也只是个小数目**，不值得为它做高成本复购激活。

### 业务行动建议（按可撬动杠杆排序）

1. **深耕"最大冷流量池"**：Organic Search 新用户桌面端做冷启动培育（再营销追回未购访客、首单礼、落地页信任模块）——**约 1,900 单增量空间，全局最大**。
2. **重仓 Referral（复购放大器）**：Referral 回访 desktop 是 ROI 最确定性组合（20.42%）；其新客 desktop 也达 10.93%，是**唯一新客跨渠道显著破 10%** 的高信任入口。做老客推荐双返、会员专属。
3. **新客信任基建**：直接访问/自然搜索的新客转化普遍只有回访一半，缺"第一单信任"。全站加强商品评价、售后承诺、资质展示。
4. **移动端降摩擦**：移动端整体约桌面端一半，且结账→购买环节折损最重（如 Organic×新用户 mobile 仅 26.5% vs 桌面 38.9%）。做免密快捷结账、一键购买，mob 聚焦低客单冲单而非高成本获客。
5. **收紧伪优质渠道**：Affiliates 加购虚火、结账弃单，Social 低意图、新客近 0 转化——这两类应按**最终转化而非加购率**考核，用券码门槛+防刷；Social 转做再营销/品牌曝光而非转化入口。
6. **警惕过度推断**：以上为观察性组合数据，"新客激活后能翻 2~3 倍"含选择性偏差（回访用户本身是自选择高意向人群），不能直接当作因果杠杆；"桌面=高客单比价"等机制解释本表无客单价支撑，仅作假设。效果需经 A/B 实验验证。

**边界**：本节结论基于观察性交叉数据，用于提出可验证的运营假设；小样本组合（浏览 <1,000，如 tablet、Social/Affiliates 部分格）的极端比率属噪音，已排除；页面、渠道与运营策略的效果需独立 A/B Test 或准实验确认。


## 4. 订单去重与 RFM 用户价值分层

本章将购买行为转换为订单级和用户级数据，使用 RFM 方法识别不同用户价值和运营客群。

**本章分析顺序：**

1. 识别购买事件并审计 `transactionId`。
2. 按订单 ID 去重，构建订单表。
3. 按用户汇总 Recency、Frequency 和 Monetary。
4. 基于 RFM 指标形成用户价值分层。

**核心原则：** 购买事件行数不等于订单数，订单数也不等于购买用户数；三者必须分别统计。

### 4.1 购买事件与订单标识审计

**分析目标**

识别 `actionType = 6` 的购买事件，检查其中 `transactionId` 的缺失情况，并比较购买事件行数、非空订单 ID 数和去重订单数。

**为什么需要这一步**

同一订单可能在事件表中出现多行。如果直接用购买事件行数作为订单数，会夸大交易规模，进而使 Frequency 和 Monetary 计算失真。

**对下一步的作用**

只有确认有效订单 ID 并完成去重，才能构建可靠的订单表，再按用户计算购买次数和累计消费金额。

````python
purchase_events = funnel_hits.loc[
    funnel_hits["actionType"] == 6
].copy()

purchase_event_count = len(purchase_events)
non_null_order_count = purchase_events["transactionId"].notna().sum()
unique_order_count = purchase_events["transactionId"].nunique()

order_audit = pd.DataFrame(
    {
        "metric": [
            "购买事件行数",
            "非空 transactionId 数",
            "去重订单数",
        ],
        "value": [
            purchase_event_count,
            non_null_order_count,
            unique_order_count,
        ],
    }
)

display(order_audit)
````

### 4.2 构建去重订单表

**分析目标**

以非空 `transactionId` 为订单标识，每个订单保留一行，构建订单级数据表。

**为什么需要这一步**

购买事件表是一张行为明细表，同一个订单可能对应多条购买事件。RFM 的 Frequency 和 Monetary 必须建立在订单粒度上，否则会重复计算购买次数和收入(一行 = 一笔订单)。

**本节处理规则**

- 删除空 `transactionId` 的购买事件。
- 按 `transactionId` 排序并保留每笔订单的第一条记录。
- 保留用户、Session、日期、订单收入、运费和税费字段。
- 原始购买事件表不修改，订单表作为新的分析对象保存。

**对下一步的作用**

订单表将作为用户 RFM 汇总的输入：按用户计算订单数、累计收入和最近购买日期。

````python
order_columns = [
    "transactionId",
    "fullVisitorId",
    "session_key",
    "date",
    "visitStartTime",
    "transactionRevenue",
    "transactionShipping",
    "transactionTax",
    "currencyCode",
]

orders = (
    purchase_events.loc[
        purchase_events["transactionId"].notna(),
        order_columns,
    ]
    .sort_values(["transactionId", "date", "visitStartTime"])
    .drop_duplicates("transactionId", keep="first")
    .reset_index(drop=True)
)

orders["order_date"] = pd.to_datetime(
    orders["date"].astype("string"),
    format="%Y%m%d",
)

orders["order_revenue_usd"] = (
    orders["transactionRevenue"]
    .fillna(0)
    / 1_000_000
)

orders.to_csv(
    output_table_dir / "orders.csv",
    index=False,
    encoding="utf-8-sig",
)

display(orders.head())
print("去重订单数:", len(orders))
print("订单用户数:", orders["fullVisitorId"].nunique())
print("订单收入（美元）:", orders["order_revenue_usd"].sum())
````

### 4.3 构建用户级 RFM 指标

**分析目标**

以购买用户为单位计算 RFM 三个核心指标：最近购买时间、去重订单数和累计订单收入。

**指标定义**

- `Recency`：观察期结束日距离用户最近一次购买的天数，越小越近期。
- `Frequency`：用户的去重订单数，越大表示购买频率越高。
- `Monetary`：用户累计订单收入，越大表示历史消费金额越高。

**为什么需要这一步**

订单表只能回答“每笔订单发生了什么”，不能直接回答“哪些用户值得重点运营”。RFM 将多笔订单压缩成一行用户级特征，便于比较用户价值(一行 = 一个购买用户)。

**对下一步的作用**

用户级 RFM 表将用于构建客户分层，例如高价值用户、忠诚用户、新近购买用户和风险用户，并为后续 Tableau 展示及召回策略提供输入。

````python
snapshot_date = orders["order_date"].max()

customer_rfm = (
    orders
    .groupby("fullVisitorId")
    .agg(
        last_order_date=("order_date", "max"),
        frequency=("transactionId", "nunique"),
        monetary=("order_revenue_usd", "sum"),
    )
    .reset_index()
)

customer_rfm["recency"] = (
    snapshot_date - customer_rfm["last_order_date"]
).dt.days

customer_rfm = customer_rfm[
    [
        "fullVisitorId",
        "recency",
        "frequency",
        "monetary",
        "last_order_date",
    ]
].sort_values(
    ["monetary", "frequency"],
    ascending=[False, False],
)

customer_rfm.to_csv(
    output_table_dir / "customer_rfm.csv",
    index=False,
    encoding="utf-8-sig",
)

display(customer_rfm.head(10))
print("RFM 用户数:", len(customer_rfm))
print("RFM 快照日:", snapshot_date)
````

### 4.4 RFM 分数与用户价值分层

**分析目标**

将 RFM 原始指标转换为 1 到 5 分，并根据分数和业务规则形成用户价值分层。

**评分规则**

- `R_score`：最近购买越近分数越高，因此 `recency` 越小越好。
- `F_score`：订单数越多分数越高。
- `M_score`：累计收入越高分数越高。
- `RFM_score`：三项分数之和，范围为 3 到 15。

**为什么需要这一步**

`recency`、`frequency` 和 `monetary` 的单位不同，不能直接相加比较。分位数评分可以把它们转换到同一尺度，便于识别高价值和风险客群。

**对下一步的作用**

用户分层结果将用于 Tableau 客群看板、运营策略和后续流失预测特征。这里的分层是规则型描述，不等同于用户的因果标签。

````python
customer_rfm_scored = customer_rfm.copy()

customer_rfm_scored["R_score"] = pd.qcut(
    customer_rfm_scored["recency"].rank(method="first"),
    q=5,
    labels=[5, 4, 3, 2, 1],
).astype(int)

customer_rfm_scored["F_score"] = pd.qcut(
    customer_rfm_scored["frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5],
).astype(int)

customer_rfm_scored["M_score"] = pd.qcut(
    customer_rfm_scored["monetary"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5],
).astype(int)

customer_rfm_scored["RFM_score"] = (
    customer_rfm_scored["R_score"]
    + customer_rfm_scored["F_score"]
    + customer_rfm_scored["M_score"]
)

def assign_customer_segment(row):
    if row["R_score"] >= 4 and row["F_score"] >= 4 and row["M_score"] >= 4:
        return "高价值用户"
    if row["F_score"] >= 4 and row["R_score"] >= 3:
        return "忠诚用户"
    if row["R_score"] >= 4 and row["F_score"] <= 2:
        return "新近购买用户"
    if row["R_score"] <= 2 and row["F_score"] >= 3:
        return "风险用户"
    if row["R_score"] >= 3 and row["F_score"] <= 2:
        return "潜力用户"
    return "一般用户"

customer_rfm_scored["customer_segment"] = (
    customer_rfm_scored.apply(assign_customer_segment, axis=1)
)

customer_rfm_scored.to_csv(
    output_table_dir / "customer_rfm_scored.csv",
    index=False,
    encoding="utf-8-sig",
)

display(customer_rfm_scored.head(10))
display(
    customer_rfm_scored["customer_segment"]
    .value_counts()
    .rename_axis("customer_segment")
    .to_frame("user_count")
)
````

````python
segment_summary = (
    customer_rfm_scored
    .groupby("customer_segment")
    .agg(
        user_count=("fullVisitorId", "nunique"),
        avg_recency=("recency", "mean"),
        avg_frequency=("frequency", "mean"),
        total_monetary=("monetary", "sum"),
        avg_monetary=("monetary", "mean"),
        avg_rfm_score=("RFM_score", "mean"),
    )
    .sort_values("total_monetary", ascending=False)
    .reset_index()
)

segment_summary["user_share"] = (
    segment_summary["user_count"] / segment_summary["user_count"].sum()
)
segment_summary["revenue_share"] = (
    segment_summary["total_monetary"]
    / segment_summary["total_monetary"].sum()
)

segment_summary.to_csv(
    output_table_dir / "customer_segment_summary.csv",
    index=False,
    encoding="utf-8-sig",
)

display(segment_summary)

print("客群用户数合计:", segment_summary["user_count"].sum())
print("客群收入合计:", segment_summary["total_monetary"].sum())
````

2. **收入集中于头部高层**：仅 **3.8% 的高价值用户**就贡献了 **25.2% 收入**——这是最集中的"金字塔尖"。若把口径放宽到较优质人群，高价值 + 新近购买合计 **18.9% 人数 → 44.1% 收入**；再纳入一般用户，则是 **48.2% 人数 → 69.7% 收入**（近半数人贡献约七成收入）。
3. **风险与忠诚是"量大但低效"的两大池子**：风险用户（2,281 人，22.8%）+ 忠诚用户（2,136 人，21.3%）合计 **44.1% 人数**，仅贡献 **21.2% 收入**。其中风险用户平均 **289 天未购**（全站最高），是**规模最大的静默流失人群**。
4. **高价值用户的画像特征**：最近购买 64 天（较近）、频次 2.89（最高）、客单 \$1,067（最高）——**复购 + 高客单**是价值来源。这跟第 2/3 章"回访用户价值高"的结论互相印证。

**业务含义（可落地）**

- **高价值用户（385 人，最值钱）**：重点维护，做会员专属权益、差异化服务，**避免过度依赖折扣**（他们已经用高复购高客单验证了价值）。这是全站"效率密度"最高的人群。
- **风险用户（2,281 人，正在流失但有一定价值）**：按历史消费高低分层召回，降本增效——历史高价值的优先、低价值的用低成本自动化触达。这是**规模最大的挽回空间**。
- **忠诚用户（2,136 人）**：已有多频复购但客单价低（\$67），方向是**提升客单价**（组合购买、交叉销售、会员套餐），而非单纯召回。
- **新近购买用户（1,510 人）**：刚买过、意向尚温，适合"购买后教育 + 关联推荐 + 二次购买引导"，是**从新客推向复购的最佳转化窗口**。
- **潜力 / 一般用户**：数量大、价值居中，用自动化、低成本常规运营，并观察其向高价值或风险迁移。

> **一句话**：全站收入高度集中——仅 **3.8% 的高价值用户**贡献 **25.2% 收入**，高价值+新近（18.9% 人数）贡献 44.1% 收入；而占比近半（44.1%）的风险+忠诚人群只产出约 21% 收入。**核心策略：把"量大但低效"的风险/忠诚人群往上推，同时死守高价值人口不让其流失。**


## 5. 业务洞察、策略建议与分析边界

本节汇总前面全部章节（数据审计、经营基线、渠道、漏斗、用户、设备、RFM）的发现，按不同方向做系统性总结，并转化为可执行的优化方向。

| **Social** | 25.0% | **0.3%** | 0.05% | **65.2%** | **大而极低效**：占了 1/4 流量却几乎不产生购买，且跳出率最高 |
| Affiliates | 1.8% | ~0.0% | 0.05% | 53.1% | 长尾、低价值 |

**核心发现**：
- **渠道的"规模"与"效率"严重错位**：流量的前三（Organic 42%、Social 25%、Direct 16%）没有带来对应的收入；反而是小流量的 Referral / Display 贡献了高效率。
- **Referral 是唯一的"利润中心"渠道**，但存在**来源单一风险**——其 42.3% 收入中绝大部分来自单一 `(direct)/(none)` 来源（约占全渠道 94.5%），且该来源标签实为"自引/直接"流量，**归因可靠性需谨慎**。
- **Social 是"大流量低价值"典型**：1/4 流量、最高跳出率、几乎零购买——应作为**品牌曝光/再营销渠道**而非转化入口。

### 5.3 漏斗洞察

整体严格漏斗（以发生商品浏览的 Session 为起点）：

| 阶段 | Session | 占比 | 相邻转化 | 流失 |
|---|---:|---:|---:|---:|
| 浏览 | 124,188 | 100% | — | — |
| 加购 | 44,044 | 35.5% | 35.47% | **64.53%** |
| 结账 | 17,088 | 13.8% | 38.80% | 61.20% |
| 购买 | 9,072 | 7.3% | **53.09%** | 46.91% |

- **全站最终转化率 7.31%**，累计流失 92.7%。
- **最大流失在「浏览→加购」（64.5%）**，其次「加购→结账」（61.2%）——**用户最卡的不是支付，而是"从看商品到决心加购"**，第一优先级是商品详情页的吸引力与加购引导，而非支付按钮。
- **组合层面**（渠道×用户×设备交叉）：`Referral×回访×desktop` 以 **20.42%** 最终转化居全矩阵第一（加购率 51.7% 也是最高）；`Social×新用户×mobile` 仅 **0.16%**；`Affiliates×新用户×desktop` 加购率看似正常（37.6%）但结账→购买仅 3.6%，是**"伪装优质"的弃单渠道**。

### 5.4 用户类型（新老用户）洞察

| 用户类型 | 浏览 | 购买 | 最终转化 |
|---|---:|---:|---:|
| 回访用户 | 41,932 | 5,548 | **13.23%** |
| 新用户 | 82,256 | 3,524 | **4.28%** |

- **回访用户转化率是新用户的 3.1 倍**，且在新老用户的各阶段对比中，回访几乎每一步都领先（浏览→加购 44.6% vs 30.8%、加购→结账 48.8% vs 31.4%、结账→购买 60.8% vs 44.2%）。
- **新用户的劣势是"系统性"的**：不是某一步特别差，而是**从进站到结账每一步都落后**——说明新用户缺的是**信任、购买习惯与决策信心**，而非某个单点问题。这为 RFM 分层与"复购激活"提供了强依据。

### 5.5 设备洞察

| 设备 | 浏览 | 购买 | 最终转化 |
|---|---:|---:|---:|
| desktop | 94,699 | 8,301 | **8.77%** |
| tablet | 3,646 | 122 | 3.35% |
| mobile | 25,843 | 649 | **2.51%** |

- **移动端转化率仅 2.51%，约为桌面端的 1/3.5**，且移动端折损发生在**加购→结账、结账→购买**两个流程环节（体验/表单/支付摩擦最重），移动端是典型的"**大流量、弱转化**"场景——适合做低客单、冲动型、免密快捷的转化，而非高成本获客。
- 设备维度是稳定分层因子：在主流渠道×用户类型的多数组合中，desktop 均优于 mobile。

### 5.6 RFM 用户价值洞察

| 客群 | 人数 | 人数占比 | 收入占比 | 人均收入 | 相对全站倍数 | 最近购买(天) | 频次 |

| 新近购买用户 | 1,510 | 15.1% | 18.9% | \$204 | 1.25× | 69 | 1.00 |
| 潜力用户 | 771 | 7.7% | 9.1% | \$191 | 1.18× | 185 | 1.00 |

- **高价值画像**：最近购买 64 天（近）、频次 2.89（最高）、客单 \$1,067（最高）——**复购 + 高客单**是价值来源。

### 5.7 复合洞察（多因素联合分析）

前文各项洞察分别从渠道、用户、设备、RFM 看，但**真正决定用户价值的是这些因素的组合**。本节基于用户主表（`user_master.csv`，将渠道、设备、新老、是否购买、RFM 客群合并到每个用户），从"**谁从哪来、用什么设备、成了一单还是高价值**"的联合视角分析。

**① 高价值用户的高度集中画像（渠道×设备×新老叠加）**

对高价值用户（356 人）做联合拆解：

| 维度 | 高价值用户的分布 | 结论 |
|---|---|---|
| **渠道** | Referral 197 / Organic 79 / Direct 71 | **Referral 一家占 55%** |
| **设备** | **desktop 345 / mobile 11** | **97% 来自桌面端** |
| **新老** | 新用户 322 / 回访用户 34 | **90% 是"新用户"** |

> **最具洞察的复合结论**：高价值用户**几乎只在 desktop + Referral 组合中出现**。它们虽然标为"新用户"（首次进入即达到高价值），但**不是靠复购堆出来的**——而是在**Referral 推荐、桌面深度浏览**的场景下，**一次进入就产生了高客单**。这说明**高价值不是"老客买得多"，而是"对的人在优质渠道一次性买得贵"**。

**② 渠道决定"买不买"，新老决定"买多少"（同渠道×新老购买率）**

| 渠道 | 回访用户购买率 | 新用户购买率 | 倍数 |
|---|---:|---:|---:|
| Referral | **10.80%** | **6.91%** | 1.6× |
| Display | 8.49% | 4.17% | 2.0× |
| Direct | 4.61% | 1.65% | 2.8× |
| Organic Search | 6.75% | 1.01% | 6.7× |
| Social | 0.32% | 0.03% | — |

> **复合印证**：同一渠道下，回访用户购买率普遍高于新用户（Referral 1.6×、Direct 2.8×、Organic 6.7×），**新用户是"打折后的回访"**。但**渠道的跨度远大于新老差距**——Referral 新用户购买率 6.91% 甚至高于 Direct 回访用户（4.61%）和 Organic 回访（6.75% 相近）。**渠道质量是第一道闸门，它决定了这批人"有没有购买意愿"；新老只决定"意愿有多强"。** Social 则两头都塌（回访 0.32%、新 0.03%），是渠道质量最差的。

**③ 设备是"放大器"（设备×新老购买率）**

| 设备 | 回访用户购买率 | 新用户购买率 |
|---|---:|---:|
| desktop | 8.01% | 1.59% |
| mobile | 2.45% | 0.48% |
| tablet | 3.43% | 0.63% |

> desktop 的回访购买率（8.01%）是 mobile（2.45%）的 **3.3 倍**，新用户更是 3.3 倍。**无论新老，desktop 都显著优于 mobile**——设备不是独立的"渠道"，而是**对购买意愿的放大器**：在好的渠道+老客基础上，desktop 进一步放大转化；在差的渠道+新客上，mobile 进一步压低。

**④ 联合结论：价值 = 渠道(意愿)×新老(强度)×设备(放大)**

把三个因素叠起来，可以解释全站的价值分布：

1. **Referral + desktop + 新用户 / 回访** = 全站最高价值带（Referral desktop 新用户 6.91%、回访 10.8%，高价值用户 97% 来自 desktop）。
2. **Organic/Direct + desktop + 回访** = 第二大价值带（高价值用户第二、三来源）。
3. **Social/移动 + 新用户** = 价值洼地（购买率 0.03%~0.48%，几乎不产生高价值用户）。

> **真正的运营杠杆不是单点优化，而是"把对的渠道 × 对的设备 × 对的人群"组合放大**：用 Referral 拉高品质新客，配合桌面端深浏览与信任建设；避免在 Social、移动端对低意向新客做高成本获客。

### 5.8 策略建议汇总（按杠杆排序）

1. **深挖"最大冷流量池"（Organic Search 新用户桌面端）**：基数最大、落差最大，做冷启动培育（再营销、首单礼、信任模块），约 **+1,900 单**增量空间，全局最大。
2. **重仓 Referral（复购放大器）**：Referral 回访 desktop 20.42% ROI 最确定；其新客 desktop 10.93% 也是唯一新客显著破 10% 的渠道。做老客推荐、会员专属。
3. **商品详情页优化（漏斗第一优先级）**：浏览→加购流失 64.5% 是最大瓶颈——信息完整性、价格/促销表达、信任元素、加购按钮可见性。
4. **购物车与结账流程优化**：加购→结账流失 61.2%，检查费用展示、登录要求、结算步骤。
5. **移动端降摩擦**：移动端约桌面端 1/3.5，结账环节折损最重，做免密快捷结账、一键购买。
6. **新客信任基建 + 复购激活**：新用户系统性劣势、回访价值高，把"拉新→复购"作为增长重心，而非纯拉新。
7. **收紧伪优质渠道**：Affiliates 加购虚火、Social 低意图且新客近零转化，按**最终转化而非加购率**考核，用券码门槛+防刷。

### 5.9 分析边界

- **观察性、非因果**：渠道差异、漏斗流失只能说明现象，不能直接证明因果关系；"新客激活后可翻 2~3 倍"等判断含选择性偏差（回访用户本身是高意向自选择人群），不应作因果杠杆。
- **统计口径**：Session 购买率、用户购买率、订单数是不同口径，不能互相替代。订单收入来自去重后的 `transactionId`，与 Session 表收入口径存在数据粒度差异。
- **RFM 相对标签**：RFM 分层基于当前观察期内的相对分位数，属规则型画像，不是绝对价值标签；小样本组合（浏览 <1,000，如 tablet / Social / Affiliates 部分客组）的极端比率属噪音，已谨慎处理。
- **需实验验证**：后续页面、渠道、运营策略应通过独立 A/B Test 或准实验确认效果。
